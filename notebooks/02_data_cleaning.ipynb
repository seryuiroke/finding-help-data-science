# Data Cleaning

## Objective

In this notebook, I'll prepare the Google Trends dataset for analysis by checking for missing values, duplicates, and formatting issues. I'll also create a cleaned version of the dataset that will be used throughout the rest of the project.

```python
import pandas as pd
```

```python
# loading the google trends data

trends = pd.read_csv("../data/raw/google_trends.csv")

trends.head()
```

```python
# view basic information about the dataset

trends.info()
```

```python
# check for missing values

trends.isnull().sum()
```

```python
# checking for duplicate rows

trends.duplicated().sum()
```

```python
# converting the date column to a datetime format

trends["date"] = pd.to_datetime(trends["date"])

trends.info()
```

```python
# saving the cleaned dataset

trends.to_csv("../data/raw/cleaned/google_trends_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

## Data Cleaning Summary

The dataset required very little cleaning because it came directly from Google Trends.

The following steps were completed:

- Checked for missing values (none found)
- Checked for duplicate rows (none found)
- Converted the date column to datetime format
- Saved a cleaned version of the dataset for future analysis

The dataset is now ready for exploratory data analysis.
